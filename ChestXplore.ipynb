The research paper I chose and will be critically evaluating today is hosted at the following link: https://linkinghub.elsevier.com/retrieve/pii/S0092867418301545.

To explore and show the extent, well in storage of the dataset, I devised the following code cell to get the size in Bytes originally of the dataset, by iterating through every directory beloning to the dataset and appending it onto the size if it is indeed a file.

```python
# get size of the pneumonia_download directory and children
dir_size = 0

# get the path and loop through each child dir
for path, dirs, files in os.walk(pneumonia_download):

  # enum the files, and loop for each
  for f in files:

  # conjoin onto path
    fp = os.path.join(path, f)

    # check if path is a file
    if os.path.isfile(fp):

      # add the size of that file to the total containing file size data object
      dir_size += os.path.getsize(fp)
```

This code cell is to format and output the above directory size, by providing a header telling the user that I will now present the explored dataset storage size, initially in Bytes > Kilobytes > Megabytes > Gigabytes.

Where the user can easily seen the maximum storage extent of the dataset I have just retrieved and am starting to explore.

```python
# Output the file size of the dataset
print(f'The size of the entire {dataset_name} Dataset is:')

# Output in storage sizes Ascending
print(dir_size, 'Bytes', ' or') # in Bytes
print(round(dir_size / 1024), 'KB', ' or') # in KB
print(round(dir_size / (1024 * 1024)), 'MB', 'or') # in MB
print(round(dir_size / (1024 * 1024 * 1024),2), 'GB') # in GB
```

Now I will populate the above created empty dataframe with all Normal images, appending them with a Normal classifcation to identify all Normal Chest images as being normal.

```python
# for every normal training image, classify as 'Normal' Chest Image
for i in os.listdir(pneumonia_training_data_normal):

    # check if i is a jpg
    if i.endswith('.jpeg'):

      # open the current normal image with PIL
      img = Image.open(pneumonia_training_data_normal + i)

      # have the image be a vector of every pixel of the desired image
      vector = np.asarray(img)

      # add to the dataframe and classify as normal
      df.loc[len(df.index)] = [vector, 'Normal']
```

Now I will resize the images in the dataset, note the quite small size (64x64), this number I came too over several tries, I tried much larger and much smaller dimensions however 64x64 still provided ample detail to the model but also performance of which we will need much of as I will be creating many CNN improvement models and pitting them to the test. I can see that this dimension is not ideal however it is a compromise to ensure notebook stability and a reasonable compile time for future cells.

```python
# Reshape images to a consistent size (e.g., 64x64) and convert to grayscale if necessary

# Initialize an empty list to store processed images
processed_images = []

# for every image
for img in images:

    # If the chest image is in grayscale already
    if len(img.shape) == 2:

        # resize the chest image to be the data standardised 64x64 for more performant colab runtime
        resized_img = cv2.resize(img, (64, 64))

    # If the chest image is in RGB colour, convert to grayscale and resize to the, again standardised 64x64
    else:
        resized_img = cv2.resize(cv2.cvtColor(img, cv2.COLOR_BGR2GRAY), (64, 64))

    # Reshape and append to the list
    processed_images.append(resized_img.reshape(64, 64, 1))

# Convert the list of processed images to a NumPy array
images = np.array(processed_images)

# Normalize pixel values to the range [0, 1], so we are only dealing with a grayscale range of colour values as the Chest Image does not require RGB colour to make a prediction as we are dealing with X-Ray Scans
images = images / 255.0
```

After converting and assuring the label data is in a compatible state, I am now in a state where I can actually create and construct the Baseline CNN model of which I will be doning so now.

```python
# Baseline CNN Model Construction

# The Baseline CNN Model
model = keras.Sequential([
    layers.Conv2D(8, (3, 3), activation='relu', input_shape=(64, 64, 1)),  # First convolutional layer, with 8 filters
    layers.MaxPooling2D((2, 2)),  # Some max pooling to reduce the dimensions of the target image
    layers.Dropout(0.2),  # Dropout to prevent some overfitting
    #layers.BatchNormalization(),  # Batch normalization, I decided to keep the baseline simple, as adding two many layers would overcomplicate having a baseline goal
    layers.Flatten(),  # Flattening the image
    layers.Dense(16, activation='relu'),  # A dense layer
    layers.Dense(1, activation='sigmoid')  # Produce the classification output using the final dense layer
])
```

##### Justification of Baseline Layers
For my baseline I designed it around just having a good base so I can have some competant performance of which I can expand and tweak later on in the solution improvement section thus the small number of layers used. 

* Firstly the first layer is the first convulational layer, which takes in the (64x64) reshaped and grayscale chest x ray scans, it uses a filter size of (3x3) as a targetor for the feature details of the given image, and uses the 'relu' activation function as I researched as that it is efficient both in terms of imporving the learning and at compiling the model.
* The secound layer is the maxpooling layer which has a 2x2 filter size, this is introduced as it 'pools' the layers thereby removing features the model deems insignificant, it also helps in reducing overfitting in CNN models as now there is less 'useless' data that the model can be fooled and overfitted from.
* The third layer is a dropout layer that seeks to drop 20% of the model neural data, this reduces the chances of overfitting by removing some of the trained data that could be overfitting
* The fourth layer is a flatten layer that turns or 'flattens' the model from dealing with two dimensions to one from where we can make a classification prediction from
* The fifth layer is a dense layer using the relu activation function again and given 16 neurons to train from, this layer attempts to take the previous 2d layers and learn from them the relations that make either a normal or pneumonia image
* The sixth and final layer makes the prediction, with 1 neuron to provide a classification that is either normal or pneumonia (0.00 or 1.00) and uses the sigmoid activation method as it is more tailored to make classification rpedictions than relu.

While it's simple it serves its purpose as a baseline allowing for competant performance from where future solution improvement models can make improvements from.

### <a id='toc1_5_1_'></a>[**Subsection 3.1: Model Hyperparameters Tuning & Tweaking**](#toc0_)

The below model is similiar to the Baseline however several parameters have been tweaked and tuned.

```python
# Hyperparameters Tweaking the Baseline CNN Model

# Building on from my work for the Baseline, I will be tweaking and adding extra layers to the model in the aim of improving the models performance over the baseline version 
model = keras.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(64, 64, 1)), # first convolutional layer, with 32 filters using the efficient relu activation function
    layers.MaxPooling2D((2, 2)), # first max pooling layer
    layers.BatchNormalization(), # batch normalization, used as it is common practise to have in a cnn model
    layers.Conv2D(64, (3, 3), activation='relu'), # second convolutional layer, this time with 2x layers, being 64
    layers.MaxPooling2D((2, 2)), # secound and final max pooling layer
    layers.Flatten(), # flatten the image
    layers.Dense(128, activation='relu'), # a dense layer with 128 neurons
    layers.Dropout(0.5), # drop out to try and catch some overfitting
    layers.Dense(1, activation='sigmoid') # final layer and make the classification using the sigmoid activation function
])
```

##### Justification of Tweaked Layers
This model not only tweaks from the baseline, but introduces new layers of which aim in improving the performance of the model from the baseline, using new techniques such as BatchNormalization and additional max pooling layers.

* The first layer contains 3 times or triple the amount of filters in the first convulation layer from the baseline, with more filters there is more of a chance of the model picking up more details from the chest x ray scan images,
* The secound layer is the first max pooling layer, with a 2x2 filter, this layer attempts to 'pool' so that the more important details are kept but lesser significant details are removed, helping to make the model to keep up with more relevant data and potentially prevent overfitting on training on irrelevant details,
* The third layer is the first and only batch normizalation layer, which is often used to improve cnn performance as it improves training speed and cnn model stability by normalization the batch size used in the model,
* The fourth layer is another convulation layer, the secound and last one, with double the filters than the first one to try and train even more details,
* The fifth layer is the secound and last max pooling layer yet again trying to reduce overfitting by removing some uneccarsary data,
* Sixth flattens and turns the data into 1D for classification prediction,
* The seventh and first dense layer to try and learn some of the image data relations made from the 2d layers, 
* The eight is a dropout layer that randomly removes 50% of the neural data to try and get rid of some overfitting,
* The last and ninth layer turns the data into a prediction either ~0 or ~1 to determine whether or not the chest x ray scan is has pneumonia or is a normal chest and is infection free.

With the improved, enhanced and additional layers I am hoping these improve the performance and stability from the baseline.

After creating the actual tweaked model, I will now attempt to compile it with a updated compile pipeline as shown below.

In the below code cell, I first retrieve and load the MobileNetV2 and fit it to so it is compatible with my X ray lungs images, do some resizing to match compatibility with the transfer learning model, pre-process the model, and then construct the MobileNetV2 model.


```python
# Transfer Learning using the MobileNetV2 Model

# Load the MobileNetV2 model from keras
base_model = tf.keras.applications.MobileNetV2(weights='imagenet', include_top=False, input_shape=(64, 64, 3))

# resize the input images to 64x64 and convert to grayscale
input_tensor = keras.Input(shape=(64, 64, 1))
x = layers.Conv2D(3, (3, 3), padding='same')(input_tensor)  # Convert grayscale to RGB chest x eay image

# preprocess the input images for the MobileNetV2 model
base_model_input = tf.keras.applications.mobilenet_v2.preprocess_input(x)

# Freeze the layers of the base model as it is a transfer learning model and already has been trained, well on others peoples work so results may not be optimal when applied to my pneumonia prediction project
base_model.trainable = False

# MobileNetV2 Model Construction
model = tf.keras.Sequential([
    keras.Model(inputs=input_tensor, outputs=base_model(base_model_input)), # pass the model the pr processed chest x ray images, and set the output of the model to be the base model passed the pre processed input images
    tf.keras.layers.GlobalAveragePooling2D(), # add a global average pooling layer to introduce some global pooling to compress the layer data
    tf.keras.layers.Dense(1, activation='sigmoid') # add a dense layer with a sigmoid activation function to make the final prediction
])
```

```python
# Transfer Learning using the EfficientNetB0 Model

# Load the EfficientNetB0 model
base_model = tf.keras.applications.EfficientNetB0(weights='imagenet', include_top=False, input_shape=(64, 64, 3))

# define the input tensors for the model to be given first by defining the 64x64 grayscale input imagess
input_tensor = keras.Input(shape=(64, 64, 1))
x = layers.Conv2D(3, (3, 3), padding='same')(input_tensor)  # Convert grayscale to RGB for keras compatibility

# preprocess the input for EfficientNetB0 model
base_model_input = tf.keras.applications.efficientnet.preprocess_input(x) # pre-processes on the chest x ray scans/images

# EfficientNetB0 Model Construction
model = tf.keras.Sequential([
    keras.Model(inputs=input_tensor, outputs=base_model(base_model_input)), # uses keras to pass the model the input tensors aka the size and grayscale image header, along with the outputs the pre processed input x ray images
    tf.keras.layers.GlobalAveragePooling2D(), # for a more consistent/less radical learning of the images as it uses global averaging
    tf.keras.layers.Dropout(0.5), # added dropout to reduce overfitting
    tf.keras.layers.Dense(1, activation='sigmoid') # predict the classification (0/1) (Normal/Pneumonia) using a sigmoid activation function
])
```

#### Justification for Transfer Learning Models
The 3 transfer learning models, starting with this one, the Efficent Net B0 model, and continuning with the next 2, is constructed around the idea of achieving compatibility with the external models to the data I have curated within this notebook and making the transfer learning models predict either normal or pneumonia ones.

* The first layer prepares and loads the efficient net b0 model, pre processes it on the pre-processed data from my x ray lungs, 
* The secound layer applies global average pooling, removing lesser detail and averaging the pooling on the more high quality detail that actually makes a difference in the model, 
* The third layer applies a 50% dropout so that overfitting is less likely as 50 percent of the neurons in the model has been removed so if it were to overfit most of that data is now gone and it cannot overfit on it anymore,
* The fourth and last model turns the model and makes a prediction from it using the same signmoid function used in previous models above to more accurstely make the prediction as it is known to be a percentage/confidence value now.

This model was constructed to allow compatability both between the loaded keras efficient net b0 model and the input data from my notebook, paving the way for the model to make a prediction from both local and external connection.

After constructing the model, I will now compile it with the below options/parameters.

### <a id='toc1_5_4_'></a>[**Subsection 3.4: Chest Image Enhancements**](#toc0_)

The below code cells, introduces the CLAHE and histogram the openCV histogram equalization common image enhancement methods, that will hopefully enanhcement the image, along with these methods a plot will be made to show the base images along with the two new image enhancers, clearly showing the effects of the two new methods compared to the images we were dealing wtih before.

```python
# Image-enhancement Methods

# load the augmented x ray lungs data I generated and save earlier on
augmented_images = np.load('augmented_x_ray_images.npy')

# store the enhanced images
enhanced_images = []

# for every augmeneted image
for image in augmented_images:

  # Convert the image to uint8 for compatibility with histogram equalization method
  image_uint8 = (image * 255).astype(np.uint8)

  # Apply the opencv histogram equalization method to the above uint8 converted chest image
  equalized_image = cv2.equalizeHist(image_uint8)

  # revert the uint8 change
  enhanced_images.append(equalized_image / 255.0)

# Convert to numpy array
enhanced_images = np.array(enhanced_images)

# Contrast Limited Adaptive Histogram Equalization (CLAHE)
enhanced_images_clahe = []

# define clahe with the following parameters
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))

# for every augmented image
for image in augmented_images:
    # convert the image to unit8 for compatibility with clahe
    image_uint8 = (image * 255).astype(np.uint8)
    
    # apply CLAHE to the image
    enhanced_image_clahe = clahe.apply(image_uint8)
    
    # grayscale the image
    enhanced_images_clahe.append(enhanced_image_clahe / 255.0)

# Convert to numpy array
enhanced_images_clahe = np.array(enhanced_images_clahe)
```

The aforementioned plot, starting with the orginal image and then the hist equalizer image then the CLAHE enhanced image, all on one row to make it easier to see the change in methods.

```python
# Plot all three, with the original image first, the histogram equalization image second and the CLAHE-enhanced image third to visualise the differences between the original and the two enhanced image methods

# Original image
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)

# plot the original image
plt.imshow(augmented_images[0], cmap='gray')

plt.title('Original')

# Histogram equalization image
plt.subplot(1, 3, 2)

# plot the histogram equalization image
plt.imshow(enhanced_images[0], cmap='gray')

plt.title('Histogram Equalization')

# CLAHE-enhanced image
plt.subplot(1, 3, 3)

# plot the CLAHE-enhanced image
plt.imshow(enhanced_images_clahe[0], cmap='gray')

plt.title('CLAHE')

# Display the original, histogram equalized, and CLAHE-enhanced images
plt.show()
```

After previously saving & saving the models, I will create a method that will test each of the solution improved models, by making a prediction on a test image (aka the firts chest image) using the same x ray lungs x and y test data, and output it back to the user, for each of the models, the prediction value along with the prediction label as-well as the accuracy and loss values to prove that we can successfully load and test and retrieve model performance.

```python
# Testing the Loaded Solutions Improved Model

# Define a function to test the loaded model
def test_loaded_model(model, X_test, y_test):
    
    # Evaluate the model
    loss, accuracy = model.evaluate(X_test, y_test, verbose=0) # generates the loss and accuracy of the loaded model

    # Output the accuracy and loss
    print(f"Accuracy: {accuracy * 100:.2f}%") # accuracy of the loaded model
    print(f"Loss: {loss:.2f}") # loss of the loaded model

    # get a test image (the first image in the testing set)
    test_image = X_test[0].reshape(1, 64, 64, 1)
    
    # Make a prediction
    prediction = model.predict(test_image) # for the current model, predict the test image classification

    # Output the prediction
    print(f"Prediction: {prediction}")

    # Label the prediction correctly, 0 = Normal, 1 = Pneumonia
    if prediction > 0.5:
        print("Pneumonia")
    else:
        print("Normal")
        
    # print a new line for better readability
    print("\n")
```

Using the above method for testing a loaded model, I will now perform the method on the rest of the solution improved CNN models and output the prediction to the notebook.

```python
# Testing the loaded baseline model using the test_loaded_model method

# Test the loaded hyperparameter-tweaked baseline model
print("Testing the loaded Baseline Hyperparameters tweaked CNN Model:")
test_loaded_model(loaded_hyperparameters_model, X_test, y_test) # hyperparameters model

# Test the loaded transfer learning models
print("Testing the loaded MobileNetV2 Model:")
test_loaded_model(loaded_mobile_net_v2_model, X_test, y_test) # MobileNetV2

print("Testing the loaded ResNet50 Model:")
test_loaded_model(loaded_res_net_50_model, X_test, y_test) # ResNet50

print("Testing the loaded EfficientNetB0 Model:")
test_loaded_model(loaded_efficient_net_b0_model, X_test, y_test) # EfficientNetB0

# Test the loaded data augmentation model
print("Testing the loaded Data Augmentation Model:")
test_loaded_model(loaded_data_aug_model, X_test, y_test) # Data Augmentation Model
```
